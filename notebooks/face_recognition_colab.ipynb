```python
import sys
from pathlib import Path
import cv2
from src.recognizer import FaceRecognizer

known_faces = Path('data/known_faces')
recognizer = FaceRecognizer(threshold=0.45)
recognizer.load_known_faces(known_faces)
print('Loaded identities:', list(recognizer.known_embeddings))
```

```python
# Example: recognize a local test image
image_path = 'test.jpg'
image = cv2.imread(image_path)
if image is None:
    raise FileNotFoundError(f'Image not found: {image_path}')

for (x1, y1, x2, y2), name, score in recognizer.recognize_image(image):
    cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)
    cv2.putText(image, f'{name} ({score:.2f})', (x1, max(25, y1 - 10)),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)

from google.colab.patches import cv2_imshow
cv2_imshow(image)
```
